# **Build an AI Assistant (Single Agent)**

An agent that responds to user queries from its knowledge.

```python
#---------------------
# DEMO: Single Agent
#---------------------

from crewai import Agent, Task, Crew, LLM

#------------------------------------------------------------------------

# Create an LLM with a temperature of 0 to ensure deterministic outputs
# OPENAI LLMs
# llm = LLM(
#          # model="gpt-5.4-mini",
#           model="gpt-5.4-nano",
#           base_url="https://api.openai.com/v1",
#           api_key = os.environ["OPENAI_API_KEY"],
#           temperature=0.7)

# GROQ hosted LLMs
llm = LLM(
     model="llama-3.3-70b-versatile",
     base_url="https://api.groq.com/openai/v1",
     api_key=os.environ["GROQ_API_KEY"],
     temperature=0.7)



# Crete an Agent
ai_assistant = Agent(
    role="AI Assistant",
    goal="""A helpful AI assistant to respond to user queries {query}""",
    backstory="""You are a helpful AI assistant who provides answers to user queries.""",
    llm=llm,
    verbose=True
)

task = Task(
  description="""A question asked by the user {query}""",
  expected_output="A suitable response to user query",
  agent = ai_assistant
)

crew = Crew(
  agents=[ai_assistant],
  tasks=[task],
  verbose=True
)
```

```python
results = await crew.kickoff_async(inputs={"query":"Who is Mahatma Gandhi?"})

print(results)
```

# **Agent with a Tool**

An agent that uses SerperDevTool to perform real-time Google search to collect information on a given topic.

```python
#------------------
# DEMO: Using Tools
#------------------

from crewai import Agent, Task, Crew, LLM
from crewai_tools import SerperDevTool

#------------------------------------------------------------------------

# Create an LLM with a temperature of 0 to ensure deterministic outputs
# OPENAI LLMs
# llm = LLM(
#          # model="gpt-5.4-mini",
#           model="gpt-5.4-nano",
#           base_url="https://api.openai.com/v1",
#           api_key = os.environ["OPENAI_API_KEY"],
#           temperature=0.7)

# GROQ hosted LLMs
llm = LLM(
     model="llama-3.3-70b-versatile",
     base_url="https://api.groq.com/openai/v1",
     api_key=os.environ["GROQ_API_KEY"],
     temperature=0.7)


#------------------------------------------------------------------------
# Create tools
search_tool = SerperDevTool()

#------------------------------------------------------------------------

# Add tools to agent
researcher = Agent(
    role="Technology Researcher",
    goal="""Research the latest developments in {topic}""",
    backstory="""You are an experienced researcher known for thorough technical research and enlightning summary of the main technical points for a gven {topic}.""",
    llm=llm,
    tools=[search_tool],
    verbose=True
)

task = Task(
  description="""Find suitable information about {topic}""",
  expected_output="Good technical summary with insights as bullet points, followed by a summary.",
  agent = researcher
)

crew = Crew(
  agents=[researcher],
  tasks=[task],
  verbose=True
)
```

```python
results = await crew.kickoff_async(inputs={"topic": "Groq platform"})

print(results)
```

# **Multi Agent System**


## The 3-Agent Setup:

## **Researcher**: Gathers core facts and key trends.

## **Writer**: Turns raw research into a structured first draft.

## **Editor**: Refines, formats, and ensures top-tier quality.

```python
import os
from crewai import LLM, Agent, Task, Crew, Process

# 2. Instantiate the Groq model
groq_llm = LLM(
    model="llama-3.3-70b-versatile",
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0.5
)

# -------------------------------------------------------------------
# AGENTS
# -------------------------------------------------------------------

# Agent 1: Research Analyst
researcher = Agent(
    role="Senior Tech Analyst",
    goal="Identify and extract key technical insights on given topics",
    backstory="You are a meticulous researcher known for finding clear, accurate, and non-fluffy technical data.",
    llm=groq_llm,
    verbose=True
)

# Agent 2: Technical Writer
writer = Agent(
    role="Content Writer",
    goal="Transform technical research into engaging, well-structured articles",
    backstory="You are a skilled developer-turned-writer who can explain complex topics simply and cleanly.",
    llm=groq_llm,
    verbose=True
)

# Agent 3: Editor
editor = Agent(
    role="Managing Editor",
    goal="Review, refine, and polish articles for maximum clarity and impact",
    backstory="You have a keen eye for tone, structure, and readability, ensuring every piece is production-ready.",
    llm=groq_llm,
    verbose=True
)

# -------------------------------------------------------------------
# TASKS
# -------------------------------------------------------------------

# Task 1: Research Phase
task_research = Task(
    description="Analyze the advantages of Groq's LPU hardware architecture for AI agent workflows.",
    expected_output="A list of 4 key technical facts and advantages of Groq's LPUs.",
    agent=researcher
)

# Task 2: Writing Phase (automatically receives context from Task 1)
task_write = Task(
    description="Draft a 2-paragraph technical blog post explaining Groq LPUs based on the research findings.",
    expected_output="A well-written 2-paragraph article draft.",
    agent=writer
)

# Task 3: Editing Phase (automatically receives context from Task 2)
task_edit = Task(
    description="Review the article draft. Fix any awkward phrasing, ensure active voice, and format with markdown headings.",
    expected_output="A polished, ready-to-publish blog post formatted in clean Markdown.",
    agent=editor
)

# -------------------------------------------------------------------
# CREW ORCHESTRATION
# -------------------------------------------------------------------

tech_crew = Crew(
    agents=[researcher, writer, editor],
    tasks=[task_research, task_write, task_edit],
    process=Process.sequential,  # Output passes sequentially: Agent 1 -> Agent 2 -> Agent 3
    verbose=True
)
```

```python

# Execute asynchronously for Google Colab environment
result = await tech_crew.kickoff_async()

print("\n\n================ FINAL OUTPUT ================\n")
print(result)
```
